## Alternative encodings
Hvis I i har problmer med at indlæse filerne med den korrekte `encoding`, kan I prøve at ændre jeres encoding i indlæsningssekvensen. Her er navnene på andre almindelige `encodings`:

- ISO-8859-1 (Latin-1): En af de mest almindelige ældre tegnkodninger for vesteuropæiske sprog, inklusive dansk.

- ISO-8859-15 (Latin-9): En mindre opdatering af ISO-8859-1, der tilføjer eurosymbolet (€) og et par andre ændringer.

- Windows-1252: Tegnkodningen brugt af Windows til vesteuropæiske sprog, stort set kompatibel med ISO-8859-1.

- UTF-16: En Unicode-kodning, der fuldt ud understøtter danske tegn og anvendes i visse applikationer, hvor fast bredde for tegn er foretrukket.

- UTF-32: Ligner UTF-16, men bruger en fast 4-byte repræsentation for hvert tegn, nyttigt i specifikke tekniske anvendelser.

Hvis I er i tvivl om, hvilken `encoding` jeres tekster har, kan i prøve denne sekvens:

```python
import chardet

with open("Kongens_fald_prep.txt", "rb") as f:
    raw_data = f.read()
    result = chardet.detect(raw_data)
    print(result["encoding"])
```
