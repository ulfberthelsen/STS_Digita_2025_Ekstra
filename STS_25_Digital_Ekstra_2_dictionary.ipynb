## Sentimentanalyse med dictionaries

En `dictionary` i Python er en datastruktur, der gemmer information i key/value-par. Det betyder, at hvert element har en unik nøgle, som bruges til at finde værdien. Når vi looper over ´dictionaries´ er det almindelig praksis at henvise til keys som `k` og values som `v`.

```python
from collections import Counter
```

```python
txt = ["De", "har", "både", "en", "sportsvogn", "og", "en", "børnecontainer"]
dict_1 = {
    "bil": 0,
    "sportsvogn": 5,
    "børnecontainer": -5,
    "skrotbunke":-5
}
score = 0

for w in txt:
    for k, v in dict_1.items(): # læses: for hver key og value i de items (key/value-par) der er indeholdt i dict_1
        if w == k:
            score = score + v
            print("Item score:", k, v)
print("Samlet score: ", score)
            
```

## Kategorisering med dictionaries

```python
# input-tyeksten
txt = ["Mette", "ringede", "først", "til", "Nicolai", "så", "til", "Troels", "og", "først", "til", "sidst", "til", "Lars"]

# dictionary
dict_2 = {
    "Mette": "socialdemokrat",
    "Nicolai": "socialdemokrat",
    "Troels": "venstremand",
    "Lars": "moderat"
}


# loop der kategoriserer inputtekstens navne efter politisk parti
for w in txt:
    for k, v in dict_2.items():
        if w == k:
            print(k, "er", v)   
```

```python
# loop der tæller hvor mange fra hvert parti, der optræder i teksten
optælling = []
for w in txt:
    for k, v in dict_2.items():
        if w == k:
            optælling.append(v)
            
```

```python
optælling
```

```python
dict_optælling = Counter(optælling)
```

```python
dict_optælling
```

```python
dict_3 = dict(dict_optælling)
```

```python
dict_3
```

```python
for k, v in dict_3.items():
    if k == "socialdemokrat":
        print("Der omtales", v, "socialdemokrater i teksten")  
```
