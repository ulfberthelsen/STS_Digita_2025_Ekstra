# Importer de nødvendige pakker

```python
import os
import nltk
from nltk.text import Text
from nltk.probability import FreqDist
```

# Stopord

```python
from nltk.corpus import stopwords
# nltk.download('stopwords') # dette skal kun gøres en gang

print(stopwords.words('danish'))  # Viser danske stopord
```

# Definer rensefunktion

```python
def rens(text_0):
    text_1 = text_0.replace("\n"," ")
    text_2 = text_1.replace("."," ")
    text_3 = text_2.replace(","," ")
    text_4 = text_3.replace(":"," ")
    text_5 = text_4.replace("*"," ")
    text_6 = text_5.replace("–"," ")
    text_7 = text_6.replace("'"," ")
    text_8 = text_7.replace("<br>"," ")
    text_ren = text_8.replace("”"," ")
    text_lav = text_ren.lower()
    text_token = text_lav.split()
    return text_token
```

# Indlæs tekst

```python
path = os.path.join("nattergalen.txt") 

with open (path, encoding = "utf8", errors="replace") as f: 
    natRaw = f.read() 
```

```python
natRaw[0:1000]
```

```python
print(natRaw[0:1000])
```

# Rens tekst

```python
natRens = rens(natRaw)
```

```python
natRens[:20]
```

# Concordance
En concordance-analyse er en metode til at undersøge, hvordan et bestemt ord eller en sætning optræder i en tekst. Man ser på ordenes kontekst, dvs. hvilke ord der står før og efter, så man kan forstå deres betydning og brug. Det bruges bl.a. i tekstanalyse til at finde mønstre i sprogbrug. For eksempel kan man undersøge, hvordan et ord bliver brugt i forskellige tekster og se, om det har en positiv eller negativ klang afhængigt af sammenhængen.

`Text()` er en `NLTK`-funtion, en såkaldt text wrapper, der pakker vores eksempel tekst ind i en masse meta-data. Det er nådvendigt at bruge denne funktion først, inden vi bruger de andre `NLTK`-funktioner.


```python
natWrap = Text(natRens)
natWrap.concordance("kejseren", lines=10)
```

```python
natWrap.concordance("nattergalen", lines=10)
```

# Collocation

Collocation betyder, at visse ord ofte optræder sammen i sproget. For eksempel siger vi 'stærk kaffe' i stedet for 'kraftig kaffe', selvom begge ord betyder nogenlunde det samme.

```python
ordPar = natWrap.collocation_list()
```

```python
ordPar
```

# Ordfrekvens


```python
ordFrekvens_1 = FreqDist(natWrap)
```

```python
print(ordFrekvens_1.most_common(10))
```

### Fjern stopord og prøv igen

```python
natGoWords = []

for o in natRens:
    if o not in stopwords.words('danish'):
        natGoWords.append(o)
```

```python
natGoWrap = Text(natGoWords)
```

```python
ordFrekvens_2 = FreqDist(natGoWrap)
```

```python
print(ordFrekvens_2.most_common(20))
```
